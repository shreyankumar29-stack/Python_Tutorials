Dictionaries aloows us to work with key-value pairs.

```python
student = {'name': 'John Doe', 'age': 20, 'major': 'Computer Science'}
# print(student['name'])
print(student['major'])
```

```python
print(student['phone'])
```

```python
print(student.get('age'))  # Using get() method to access the value of 'age'
```

```python
print(student.get('phone'))
```

```python
print(student.get('phone', 'Phone number not available'))  # Using get() method with a default value
```

```python
student['phone'] = '123-456-7890'  # Adding a new key-value pair to the dictionary  
student['name'] = 'Jane Smith'  # Updating the value of an existing key in the dictionary
print(student)  # Printing the updated dictionary
```

```python
student.update({'age': 21, 'major': 'Data Science'})  # Updating multiple key-value pairs in the dictionary
print(student)  # Printing the updated dictionary after using the update() method
```

```python
del(student['age'])  # Deleting a key-value pair from the dictionary
print(student)  # Printing the updated dictionary after deleting a key-value pair
```

```python
age = student.get('age', 'Age not available')  # Using get() method with a default value
print(student)
print(age)
```

Count keys in a dictionary

```python
print(len(student))  # Printing the number of key-value pairs in the dictionary
```

```python
print(student.values())  # Printing all the values in the dictionary
```

```python
print(student.items())  # Printing all the key-value pairs in the dictionary as tuples
```

```python
for key in student:
    print(key)
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```
