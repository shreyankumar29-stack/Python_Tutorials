Lists:In Python, a list is a built-in data structure used to store multiple items in a single variable. Lists are ordered, mutable (changeable), and allow duplicate values.

```python
courses = ['History', 'Math', 'Physics', 'CompSci']
courses2 = ['Art', 'Education']  #changing first element of list
# courses.insert(0, 'Art')  #inserting element at index 0
# courses.append('Art')  #adding element to the end of list
# courses.insert(0, courses2)  #inserting list at index 0

courses.extend(courses2)  #adding list to the end of list
print(courses)  #printing list
print(courses[0])  #printing first element of list
print(courses[3])  #printing last element of list
print(courses[-1])  #printing last element of list using negative indexing
print(courses[:2])  #printing first two elements of list
print(courses[2:4])  #printing elements from index 2 to 3 (end index is exclusive)

```

```python
print(courses[4])    #this will give an error because index 4 does not exist in the list
```

```python
courses = ['History', 'Math', 'Physics', 'CompSci']
# courses.remove('Math')  #removing element from list
popped = courses.pop()  #removing last element from list and storing it in a variable
print(courses)  #printing list
print(popped)
```

Sorting Lists

```python
courses = ['History', 'Math', 'Physics', 'CompSci']

nums = [1, 5, 2, 4, 3]
courses.reverse()   #lists in reverse order
courses.sort()     #lists in ascending order
nums.sort()        #lists in ascending order
courses.sort(reverse=True)  #lists in descending order
nums.sort(reverse=True)  #lists in descending order
print(nums)
print(courses)
```

```python
courses = ['History', 'Math', 'Physics', 'CompSci']

nums = [1, 5, 2, 4, 3]

sorted_courses = sorted(courses)  #returns a new list in ascending order without changing the original list
sorted_nums = sorted(nums)        #returns a new list in ascending order without changing the original list

print(sorted_courses)
print(sorted_nums)
```

Built-in Functions:


```python
courses = ['History', 'Math', 'Physics', 'CompSci']

nums = [1, 5, 2, 4, 3]
sorted_courses = sorted(courses)  #returns a new list in ascending order without changing the original list
sorted_nums = sorted(nums)        #returns a new list in ascending order without changing the original

print(sorted_courses)
print(max(sorted_nums))
print(min(sorted_nums))
print(sum(sorted_nums))
```

Defining the value of a certain index


```python
courses = ['History', 'Math', 'Physics', 'CompSci']
print(courses.index('Math'))  #returns the index of the element in the list
print('Art'in courses)  #returns True if the element is in the list, False otherwise    
```

```python
print(courses.index('Art'))  #this will give an error because 'Art' is not in the list
```

```python
for item in courses:
    print(item)
```

enumerate functions:


```python
courses = ['History', 'Math', 'Physics', 'CompSci']
for index, course in enumerate(courses, start=1):
    print(index, course)
```

Join

```python
courses = ['History', 'Math', 'Physics', 'CompSci']

course_str =' - '.join(courses)
new_list = course_str.split(' - ')  #splitting string into a list
print(new_list)
print(course_str)  #printing list as a string
```
