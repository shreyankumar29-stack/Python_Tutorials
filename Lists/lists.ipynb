Lists:In Python, a list is a built-in data structure used to store multiple items in a single variable. Lists are ordered, mutable (changeable), and allow duplicate values.

```python
courses = ['History', 'Math', 'Physics', 'CompSci']
courses2 = ['Art', 'Education']  #changing first element of list
courses.insert(0, 'Art')  #inserting element at index 0
# courses.append('Art')  #adding element to the end of list
courses.insert(0, courses2)  #inserting list at index 0
print(courses)  #printing list
print(courses[0])  #printing first element of list
print(courses[3])  #printing last element of list
print(courses[-1])  #printing last element of list using negative indexing
print(courses[:2])  #printing first two elements of list
print(courses[2:4])  #printing elements from index 2 to 3 (end index is exclusive)

```
