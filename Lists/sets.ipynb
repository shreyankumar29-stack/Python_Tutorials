Sets: It is used to test whether a value is part of a set or not

```python
cs_sets = {'History', 'Math', 'Physics', 'CompSci','Math'}

print('Math' in cs_sets)
```

```python
cs_sets = {'History', 'Math', 'Physics', 'CompSci','Math'}
arts_sets = {'History', 'Math', 'Art', 'Design'}

print(arts_sets.intersection(cs_sets))
print(cs_sets.difference(arts_sets))
print(cs_sets.union(arts_sets))
```
