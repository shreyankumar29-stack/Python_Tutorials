Tuples: Similar to lists but we can't modify tuples(immutable)

```python
# Mutable

# list_1 = ['History', 'Math', 'Physics', 'CompSci']
# list_2 = list_1  #this will create a reference to the original list, not a copy

# print(list_1)
# print(list_2)

# list_1[0] = 'Art'

# print(list_1)
# print(list_2)  #this will also change because list_2 is a reference to list_1  


# Immutable

tuple_1 = ('History', 'Math', 'Physics', 'CompSci')
tuple_2 = tuple_1  #this will create a reference to the original tuple, not a copy

print(tuple_1)
print(tuple_2)
```

```python
tuple_1[0] = 'Art'  #this will give an error because tuples are immutable   
```
